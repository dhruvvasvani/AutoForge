```python
from pathlib import Path
import sys
import json
import torch
import pandas as pd

# 1. Setup paths
analyzer = Path.cwd()
if analyzer.name != "3_ast_analyzer":
    analyzer = analyzer / "3_ast_analyzer"

orig_train = analyzer / "data" / "pycode_vul" / "prepared" / "train.csv"
if not orig_train.exists():
    orig_train = analyzer / "data" / "pycode_vul" / "train.csv"
orig_test = analyzer / "data" / "pycode_vul" / "test.csv"

# 2. Create micro-subsets for instant demo execution
demo_dir = analyzer / "data" / "pycode_vul" / "demo"
demo_dir.mkdir(parents=True, exist_ok=True)

demo_train = demo_dir / "train.csv"
demo_test = demo_dir / "test.csv"

# Sample 300 train and 100 test items
pd.read_csv(orig_train).head(300).to_csv(demo_train, index=False)
pd.read_csv(orig_test).head(100).to_csv(demo_test, index=False)

sys.path.insert(0, str(analyzer))
from src.ml.cpg_training import train_cpg_gnn

checkpoint = analyzer / "src" / "ml" / "cpg_gnn_model.pt"
results_dir = analyzer / "results" / "week7_gnn"
results_dir.mkdir(parents=True, exist_ok=True)

print("Starting Fast Demo Training (Micro Subset)...")
result = train_cpg_gnn(
    dataset_path=demo_train,
    test_dataset_path=demo_test,
    checkpoint_path=checkpoint,
    max_epochs=2,
    patience=7,
    seed=42,
)

# 3. Save metrics artifact
metrics_path = results_dir / "training_metrics.json"
metrics_path.write_text(json.dumps(result, indent=2), encoding="utf-8")

print("\nSUCCESS! Artifacts Generated:")
print("1. Checkpoint:", checkpoint.exists(), checkpoint)
print("2. Metrics:", metrics_path.exists(), metrics_path)
```
