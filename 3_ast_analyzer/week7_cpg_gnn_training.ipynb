```python
from pathlib import Path
import sys
import json
import torch

# 1. Automatically find project root
analyzer = Path.cwd()
if analyzer.name != "3_ast_analyzer":
    analyzer = analyzer / "3_ast_analyzer"

train_csv = analyzer / "data" / "pycode_vul" / "prepared" / "train.csv"
if not train_csv.exists():
    train_csv = analyzer / "data" / "pycode_vul" / "train.csv"
test_csv = analyzer / "data" / "pycode_vul" / "test.csv"

sys.path.insert(0, str(analyzer))
from src.ml.cpg_training import train_cpg_gnn

# 2. Setup output paths
checkpoint = analyzer / "src" / "ml" / "cpg_gnn_model.pt"
results_dir = analyzer / "results" / "week7_gnn"
results_dir.mkdir(parents=True, exist_ok=True)

# 3. Train fast GNN model (2 Epochs)
print("Starting Training...")
result = train_cpg_gnn(
    dataset_path=train_csv,
    test_dataset_path=test_csv,
    checkpoint_path=checkpoint,
    max_epochs=2,
    patience=7,
    seed=42,
)

# 4. Save results metrics
metrics_path = results_dir / "training_metrics.json"
metrics_path.write_text(json.dumps(result, indent=2), encoding="utf-8")

print("\nSUCCESS!")
print("Checkpoint saved at:", checkpoint)
print("Metrics saved at:", metrics_path)
```
