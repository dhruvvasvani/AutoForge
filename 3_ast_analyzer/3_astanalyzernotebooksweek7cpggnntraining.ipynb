```python
import os
import pandas as pd

possible_paths = [
    '3_ast_analyzer/data/pycode_vul/train.csv',
    '../data/pycode_vul/train.csv',
    'data/pycode_vul/train.csv'
]

train_path = None
for path in possible_paths:
    if os.path.exists(path):
        train_path = path
        break

if train_path:
    test_path = train_path.replace('train.csv', 'test.csv')
    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)
    print(f"Loaded from: {train_path}")
    print(f"Train samples: {len(train_df)}")
    print(f"Test samples: {len(test_df)}")
    print("\nTrain Class Distribution:\n", train_df['label'].value_counts())
else:
    print("Error: CSV files nahi mili. Current directory path:", os.getcwd())
```
